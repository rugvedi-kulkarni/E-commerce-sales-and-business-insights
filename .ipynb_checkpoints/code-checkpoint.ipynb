```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset/ecommerce_retail_seaborn_practice_100k.csv")

df["order_date"] = pd.to_datetime(df["order_date"])
```

Question: What business problem am I trying to understand?

Visualization: Which Seaborn plot is appropriate, and why?

Observation: What pattern can I actually see?

Insight: What does that pattern mean?

Action: What decision could someone make from this finding?

```python
sns.set_theme(style = "darkgrid")
```

```python

print("Original shape:", df.shape)

# sample dataset

sample_df = df.sample(
    n=min(5000, len(df)),
    random_state=42
)

print("Sample shape:", sample_df.shape)
```

```python
#  Is there a relationship between units sold and sales revenue?
sns.set_theme(style = "whitegrid")

sns.lineplot(
    data = sample_df, 
    x = "sales_revenue_inr",
    y = "units_sold",
    color = "green",
    label = "increase in sales revenue"
)

plt.title("Units sold for sales revenue")
plt.xlabel("Sales revenue {in inr}")
plt.ylabel("Units sold")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# How does unit price relate to profit across different product categories?
sns.scatterplot(
    data = sample_df,
    x = "unit_price_inr",
    y = "profit_inr",
    hue = "product_category"
)

plt.title("Distribution of unit price and profit across product category")
plt.xlabel("Unit price {in inr}")
plt.ylabel("Profit {in inr}")
plt.tight_layout()
plt.show()
```

```python
# How does the discount percentage relate to profit?

sns.barplot(
    data = sample_df,
    x = "discount_pct",
    y = "profit_inr"
)

plt.title("Effect of discount % on profit")
plt.xlabel("Discount %")
plt.ylabel("Profit {in inr}")
plt.tight_layout()
plt.show()
```

```python
# Does the relationship between sales revenue and profit differ by product category?

sns.histplot(
    data = sample_df, 
    x = "sales_revenue_inr",
    y = "profit_inr",
    bins = 30,
    hue = "product_category"
)

plt.title("Sales revenue vs profit for different product category")
plt.xlabel("Sales revenue {in inr}")
plt.ylabel("Profit {in inr}")
plt.tight_layout()
plt.show()
```

```python
# How do customer ratings vary with sales revenue?

sns.lineplot(
    data = sample_df,
    x = "customer_rating",
    y = "sales_revenue_inr"
)

plt.title("Customer ratings for sales revenue")
plt.xlabel("Customer rating (1 - 5)")
plt.ylabel("Sales revenue {in inr}")
plt.tight_layout()
plt.show()
```

```python
# How do shipping days relate to customer ratings?
# Do orders with longer shipping times show different rating patterns?

sns.set_theme(style = "whitegrid")

sns.lineplot(
    data = sample_df,
    x = "shipping_days",
    y = "customer_rating",
    errorbar = None,
    color = "green",
    linewidth = 1.5 
)

plt.title("variation of ratings on shipment time")
plt.xlabel("Shipping days")
plt.ylabel("Customer rating (1 - 5)")
plt.tight_layout()
plt.show()
```

```python
# How do customer ratings vary across different sales channels?

sns.barplot(
    data = sample_df,
    x = "customer_rating",
    y = "sales_channel" 
)

plt.title("customer ratings for sales channel")
plt.xlabel("Customer rating (1 - 5)")
plt.ylabel("Sales channel")
plt.tight_layout()
plt.show()
```

```python
# How do shipping days relate to sales revenue?

ax = sns.barplot(
    data = sample_df,
    x = "shipping_days",
    y = "sales_revenue_inr",
    color = "orange"
)


plt.title("Trends in sales revenue according to shipment time")
plt.xlabel("Shipping days")
plt.ylabel("Sales revenue {in inr}")
plt.tight_layout()
plt.show()
```

```python
# How do customer ratings vary with discount percentage?

sns.set_theme(style = "whitegrid")

sns.lineplot(
    data = sample_df,
    x = "discount_pct",
    y = "customer_rating",
    color = "purple",
    errorbar = None,
    marker = "o",
    markersize = 8.5,
    label = "change in rating"
)



plt.title("Effect of discount % on customer ratings")
plt.xlabel("Discount %")
plt.ylabel("Customer rating")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
 # Do orders with longer shipping times show different rating patterns?

sns.set_theme(style = "ticks")

sns.histplot(
    data = sample_df,
    x = "shipping_days",
    y = "customer_rating",
    color = "green",
    bins = 5 
)

plt.title("Effect of shipping days on rating")
plt.xlabel("Shipping days")
plt.ylabel("Customer rating")
plt.tight_layout()
plt.show()
```

```python
# How do unit prices vary across product categories?
sns.set_theme(style = "ticks")

sns.histplot(
    data = sample_df,
    x = "product_category",
    y = "unit_price_inr",
    bins = 10,
    
)

plt.title("Price variation across categories")
plt.xlabel("Product category")
plt.ylabel("Unit price {in inr}")
plt.tight_layout()
plt.show()
```

```python
# How do sales revenues vary across customer segments?

sns.barplot(
    data = sample_df,
    x = "customer_segment",
    y = "sales_revenue_inr",
    label = "revenue"
)


plt.title("Sales revenue for different customers segments")
plt.xlabel("Customer segment")
plt.ylabel("Sales revenue {in inr}")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#  How do sales revenues differ across sales channels?
sns.set_theme(style = "ticks")

sns.barplot(
    data = sample_df,
    # kind = "line"
    x = "sales_revenue_inr",
    y = "sales_channel",
    color = "orange"
)

plt.title("Change in sales revenue for different sales channels")
plt.xlabel("Sales revenue {in inr}")
plt.ylabel("Sales channel")
plt.tight_layout()
plt.show()
```

```python
# How do discounts vary across product categories?

each_percent = sns.lineplot(
    data = sample_df,
    x = "product_category",
    y = "discount_pct",
    errorbar = None,
    marker = "o",
    linewidth = 1.5,
    color = "green"
    
)

    
plt.title("Discount vary across product category")
plt.xlabel("Product category")
plt.ylabel("Discount per %")
plt.tight_layout()
plt.show()
```

```python
# How do sales revenue and profit relate across different regions?

ax = sns.histplot(
    data = sample_df,
    x = "sales_revenue_inr",
    y = "profit_inr",
    hue = "product_category",
    bins = 20,
)

plt.title("Sales revenue and profit across different product category")
plt.xlabel("Sales revenue {in inr}")
plt.ylabel("Sales Profit {in inr}")
plt.tight_layout()
plt.show()
```

```python
 # How do shipping days vary across regions?
sns.set_theme(style = "white")

sns.barplot(
    data  = sample_df,
    x = "region",
    y = "shipping_days",  
)

plt.title("Shipping days for regions")
plt.xlabel("Regions")
plt.ylabel("Shipping days")
plt.tight_layout()
plt.show()
```

```python
 # 18. How does sales revenue change over time?
sns.scatterplot(
    data = sample_df,
    x = "order_date",
    y = "sales_revenue_inr",
)

plt.title("Change in sales revenue over time")
plt.xlabel("Time {in terms of date}")
plt.ylabel("Sales revenue {in inr}")
plt.tight_layout()
plt.show()
```
